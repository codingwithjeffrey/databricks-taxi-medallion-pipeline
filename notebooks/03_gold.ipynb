```python
spark.sql("USE workspace.taxi_medallion")

spark.sql("""
CREATE OR REPLACE TABLE gold_daily_zip_revenue AS
SELECT pickup_date, pickup_zip,
       COUNT(*) AS trips,
       ROUND(SUM(fare_amount), 2) AS total_fare,
       ROUND(AVG(trip_distance), 2) AS avg_distance,
       ROUND(AVG(trip_minutes), 1) AS avg_minutes
FROM silver_trips
GROUP BY pickup_date, pickup_zip
""")

spark.sql("""
CREATE OR REPLACE TABLE gold_hourly_demand AS
SELECT pickup_hour, COUNT(*) AS trips, ROUND(AVG(fare_amount), 2) AS avg_fare
FROM silver_trips
GROUP BY pickup_hour
""")

display(spark.table("gold_hourly_demand").orderBy("pickup_hour"))
```
